## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

### Merging Data

Setelah melewati proses pembersihan, kita kini memiliki tiga sumber data utama yang sudah siap: `df_orders` yang berisi informasi status dan riwayat waktu pesanan, `df_customers` yang memuat profil dan demografi pelanggan, serta `df_payments` yang berisi rincian nilai dan metode pembayaran.

Pada tahap ini, kita akan menggabungkan ketiga dataset tersebut secara bertahap menjadi satu DataFrame utama (kita sebut `df_main`) menggunakan *method* `.merge()` bawaan dari library `pandas`.

Kita menggunakan pendekatan `INNER JOIN` untuk memastikan bahwa hanya data pesanan yang memiliki informasi pelanggan dan catatan pembayaran yang lengkap saja yang akan diikutsertakan dalam analisis final.

**Logika Penggabungan:**
* **Metode:** `INNER JOIN` (parameter `how='inner'`).
* **Tahap 1 (Orders & Customers):** Menggabungkan tabel `df_orders` sebagai tabel utama dengan `df_customers` menggunakan kunci penghubung (*Primary Key*) berupa kolom `customer_id`.
* **Tahap 2 (Menambahkan Payments):** Menggabungkan DataFrame hasil dari Tahap 1 dengan tabel `df_payments` menggunakan kunci penghubung berupa kolom `order_id`.

```python
# Menggabungkan Orders dengan Customers
df_merged = df_orders.merge(df_customers, on='customer_id', how='inner')

# Menggabungkan hasil di atas dengan Payments untuk mendapatkan DataFrame Utama
df_main = df_merged.merge(df_payments, on='order_id', how='inner')

print(f"Total baris data bersih yang siap dianalisis: {df_main.shape[0]}")
```

## Exploratory Data Analysis (EDA)

### Tujuan utama dari tahap ini adalah untuk memahami distribusi data, mendeteksi korelasi antar variabel, dan untuk menjawab keempat pertanyaan bisnis.

```python
df_main.describe()
```

```python
# 1. Menyiapkan Data Waktu (Untuk Pertanyaan 1)
df_main['purchase_day'] = df_main['order_purchase_timestamp'].dt.day_name()
df_main['purchase_hour'] = df_main['order_purchase_timestamp'].dt.hour
```

```python
# 2. Menyiapkan Data RFM (Untuk Pertanyaan 4)
recent_date = df_main['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

rfm_df = df_main.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (recent_date - x.max()).days,
    'order_id': 'nunique',
    'payment_value': 'sum'
}).reset_index()
rfm_df.columns = ['customer_unique_id', 'Recency', 'Frequency', 'Monetary']
```

```python
# Pembuatan Segmen RFM Manual
rfm_df['R_Score'] = pd.qcut(rfm_df['Recency'], 4, labels=[4, 3, 2, 1])
rfm_df['F_Score'] = rfm_df['Frequency'].apply(lambda x: 4 if x > 2 else (3 if x == 2 else 1))
rfm_df['M_Score'] = pd.qcut(rfm_df['Monetary'], 4, labels=[1, 2, 3, 4])

def segment_customer(df):
    if df['R_Score'] == 4 and df['F_Score'] >= 3 and df['M_Score'] >= 3: return 'Champions'
    elif df['R_Score'] >= 3: return 'Recent Customers'
    elif df['R_Score'] == 1 and df['F_Score'] == 1: return 'Lost Customers'
    else: return 'Regular Customers'

rfm_df['Customer_Segment'] = rfm_df.apply(segment_customer, axis=1)
```

```python
# Mencari hubungan linear antar semua variabel numerik
corr_matrix = df_main.corr(numeric_only=True)

print("--- Korelasi Fitur Numerik terhadap Payment Value ---")
kolom = ['payment_value', 'payment_installments', 'payment_sequential', 'purchase_hour']
print(corr_matrix.loc[kolom, 'payment_value'].sort_values(ascending=False))
```

## Analisis Lanjutan

```python
recent_date = df_main['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

# Menghitung Recency, Frequency, dan Monetary (RFM)
rfm_df = df_main.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (recent_date - x.max()).days, # Recency: Selisih hari
    'order_id': 'nunique',                                              # Frequency: Jumlah transaksi
    'payment_value': 'sum'                                              # Monetary: Total pengeluaran
}).reset_index()

rfm_df.columns = ['customer_unique_id', 'Recency', 'Frequency', 'Monetary']

rfm_df['R_Score'] = pd.qcut(rfm_df['Recency'], 4, labels=[4, 3, 2, 1])
rfm_df['F_Score'] = rfm_df['Frequency'].apply(lambda x: 4 if x > 2 else (3 if x == 2 else 1))
rfm_df['M_Score'] = pd.qcut(rfm_df['Monetary'], 4, labels=[1, 2, 3, 4])

def segment_customer(df):
    if df['R_Score'] == 4 and df['F_Score'] >= 3 and df['M_Score'] >= 3:
        return 'Champions'
    elif df['R_Score'] >= 3:
        return 'Recent Customers'
    elif df['R_Score'] == 1 and df['F_Score'] == 1:
        return 'Lost Customers'
    else:
        return 'Regular Customers'

rfm_df['Customer_Segment'] = rfm_df.apply(segment_customer, axis=1)

print(rfm_df['Customer_Segment'].value_counts())
rfm_df.head()
```

Penempatan inventaris yang terpusat di SP akan memangkas biaya ongkos kirim secara drastis dan mempercepat waktu pengiriman (SLA Delivery) ke mayoritas pelanggan, yang pada akhirnya akan meningkatkan Review Score platform.

- Strategi untuk Pelanggan Baru (CRM/Retensi): Karena platform memiliki jumlah "Recent Customers" yang sangat besar, Tim Customer Relationship Management (CRM) wajib merancang program untuk mendorong pembelian kedua. Kirimkan email atau notifikasi berisi Voucher Spesial Pesanan Kedua seminggu setelah barang pertama mereka sampai. Tujuannya adalah mengonversi para "Recent Customers" ini agar naik kelas menjadi "Champions" sebelum mereka berubah menjadi "Lost Customers".
